## Overview

DMD (via pyDMD) is implemented in following steps:

- initialise a DMD class object
- (optional) select hankel delay time embedding (d) and apply 
- fit to the time series data (X)

## Initialising DMD class objects

```python
def Build_Exact_DMD(svd_rank=-1):

    return DMD(
        svd_rank=svd_rank,
        exact=True,
        forward_backward=False,
    )

def Build_BOPDMD(svd_rank=-1):

    return BOPDMD(
        svd_rank=svd_rank,
    )

dmd_base = Build_Exact_DMD()
```

## Hankel preprocessing

```python
hankel=False

if hankel:
    d_choice = 3
    dmd = hankel_preprocessing(dmd_base, d=d_choice, reconstruction_method="first")
else:
    dmd = dmd_base
```
