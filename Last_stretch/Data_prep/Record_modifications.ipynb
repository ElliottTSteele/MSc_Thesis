# Record Modifications

Notebook implementing the modifications made to the record used for DMD input.

Full record includes:

- Spatial degree of $N_{max}=20$
    - This should always be truncated to 15 or lower in line with the CHAOS-8 author recommendations

- Record time length of 29.4 years (1997.1-2026.5)
    - This must be truncated to 26 years (2000.1-2026.1) 

- Record sample spacing uses 0.2 years from resolution output
    - This is extended to 0.4 years

## Applying modifications to record

```python
# takes in gauss time series (shape=(148, 440)) over full record and prepares
# its spatial and temporal information
def Gauss_Record_Prepare(gauss_full, tmax=15, t_sample_rate=2):

    # truncate
    gauss_truncated = Truncate_Gauss_Coeffs(gauss_full, tmax)

    # window in time
    gauss_windowed = gauss_truncated[t_window_slice, :]

    # downsample in time from 0.2 to 0.4 year sample interval
    gauss = gauss_windowed[::t_sample_rate, :]

    return(gauss)
```
